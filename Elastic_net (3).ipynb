```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,ElasticNet
```

```python
df=pd.read_excel(r"C:\Users\CHAIT\OneDrive\Desktop\TE\Machine Learning\Dataset\Supervised\demo\elastic_net_demo_dataset.xlsx")
df.head()
```

```python
df=df.drop(columns=['Employee_ID','City'],axis=1)
```

```python
df.head()
```

```python
df["Education"].describe()
```

```python
df["Department"].describe()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
one=OneHotEncoder(sparse_output=False)
```

```python
encode=one.fit_transform(df[['Education','Department']])
```

```python
print(encode)
```

```python
encoded_df=pd.DataFrame(encode,columns=one.get_feature_names_out(['Education','Department']))
encoded_df
```

```python
df=pd.concat([df,encoded_df],axis=1)
```

```python
df.head()
```

```python
df=df.drop(['Education','Department'],axis=1)
```

```python
df.head()
```

```python
x=df.drop(columns=['Annual_Bonus'],axis=1)
```

```python
y=df['Annual_Bonus']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.fit_transform(x_test)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
```

```python
le=LinearRegression()
le.fit(x_train,y_train)
```

```python
y_pred = le.predict(x_test)

print("MSE =", mean_squared_error(y_test, y_pred))
print("Mae =", mean_absolute_error(y_test, y_pred))
print("r2_score=",r2_score(y_test,y_pred)*100)
```

```python
e=ElasticNet(alpha=1.0,l1_ratio=1)
e.fit(x_train,y_train)
```

```python
y_pred = e.predict(x_test)

print("MSE =", mean_squared_error(y_test, y_pred))
print("Mae =", mean_absolute_error(y_test, y_pred))
print("r2_score=",r2_score(y_test,y_pred)*100)
```
