```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans,DBSCAN
```

```python
x,y_true=make_moons(n_samples=10000,noise=0.05,random_state=42)
df=pd.DataFrame(x,columns=['f1','f2'])
```

```python
km=KMeans(n_clusters=2)
db=DBSCAN(eps=0.5,min_samples=5)
```

```python
sc=StandardScaler()
x_sc=sc.fit_transform(x)
```

```python
km.fit(x_sc)
```

```python
km_labels=km.predict(x_sc)
df['km_labels']=km_labels
sns.scatterplot(x=df['f1'],y=df['f2'],hue=df['km_labels'])
```

```python
db_labels=db.fit(x_sc)
```

```python

df['db_labels']=db_labels
sns.scatterplot(x=df['f1'],y=df['f2'],hue=df['db_labels'])
```
