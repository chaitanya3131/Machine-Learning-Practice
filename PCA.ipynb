```python
import pandas as pd
import time
from sklearn.datasets import load_diabetes
from sklearn.metrics import mean_squared_error,r2_score
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data=load_diabetes(as_frame=True)
```

```python
x=data.data
y=data.target
```

```python
df=pd.concat([x,y],axis=1)
```

```python
pca=PCA(n_components=0.95)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

## MODEL WITHOUT PCA

```python
s=time.time()
li=LinearRegression()
li.fit(x_train,y_train)
e=time.time()
print('Traing time= ',s-e)
print('mse= ',mean_squared_error(y_test,li.predict(x_test)))
```

## MODEL WIHT PCA

```python
x_train_pca=pca.fit_transform(x_train)
x_test_pca=pca.transform(x_test)
```

```python
s=time.time()
lipc=LinearRegression()
lipc.fit(x_train_pca,y_train)
e=time.time()
print('Traing time= ',e-s)
print('mse= ',mean_squared_error(y_test,lipc.predict(x_test_pca)))
```
