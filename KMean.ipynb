```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
x,y_true=make_blobs(n_samples=10000,centers=3,random_state=42,cluster_std=0.5)
```

```python
df=pd.DataFrame(x,columns=['f1','f2'])
```

```python
sc=StandardScaler()
df_scaled=sc.fit_transform(df)
```

```python
krange=range(1,11)
inertia=[]
```

```python
for k in krange:
    kmeans=KMeans(n_clusters=k)
    kmeans.fit(x)
    inertia.append(kmeans.inertia_)
    
```

```python
inertia
```

```python
plt.plot(krange,inertia,marker='o')
```

```python
f_km=KMeans(n_clusters=3)
```

```python
f_km.fit(df_scaled)
```

```python
cluster_labels=f_km.predict(df_scaled)
df['clusters']=cluster_labels
```

```python
sns.scatterplot(x=df['f1'],y=df['f2'],hue=df['clusters'])
```
