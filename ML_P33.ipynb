```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\CHAIT\OneDrive\Desktop\ML_git_hub\california_housing.csv")
```

```python
df.head()
```

```python
x=df.drop(columns=['MedHouseVal'],axis=1)
```

```python
y=df['MedHouseVal']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)



li=LinearRegression()
li.fit(x_train,y_train)
```

```python
round(),2)
```

```python
plt.figure(figsize=(8,4))

plt.plot(
    ['Training Error', 'Testing Error'],
    [
        mean_squared_error(y_train, li.predict(x_train)),
        mean_squared_error(y_test, li.predict(x_test))
    ]
    ,marker='*',linestyle='--'
)

plt.show()
```
