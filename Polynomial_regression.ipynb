```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing,load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
data=load_diabetes(as_frame=True)
x=data.data
y=data.target
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python

plt.scatter(y,y)
plt.scatter(y,lr.predict(x),color='red')
plt.show()
```

```python
r2_score(y,lr.predict(x))
```

```python
from sklearn.preprocessing import PolynomialFeatures
p=PolynomialFeatures(degree=2)
x_train_poly=p.fit_transform(x_train)
x_test_poly=p.fit_transform(x_test)
```

```python
lr2=LinearRegression()
lr2.fit(x_train_poly,y_train)
```

```python

x_poly=p.transform(x)
plt.scatter(y,y)
plt.scatter(y,lr2.predict(x_poly),color='red')
plt.show()
```

```python
r2_score(y,lr2.predict(x_poly))
```

```python
from sklearn.linear_model import Ridge,Lasso
```

```python
r=Ridge(alpha=1)
r.fit(x_train,y_train)
```

```python
r2_score(y,r.predict(x))
```

```python
from sklearn.tree import DecisionTreeRegressor

t=DecisionTreeRegressor()
t.fit(x_train,y_train)
```

```python
r2_score(y,t.predict(x))
```
