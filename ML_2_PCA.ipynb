```python
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
```

```python
data=load_diabetes()
df=pd.DataFrame(data.data,columns=data.feature_names)
df['target']=data.target
df.head()
```

```python
x=df.drop(columns='target',axis=1)
y=df['target']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

## MODEL BEFORE THE PCA

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
le_without_pca=LinearRegression()
```

```python
import time
```

```python
start=time.time()
le_without_pca.fit(x_train,y_train)
end=time.time()
print("The time required for traing is : ",start-end)
```

```python
y_pred_without_pca=le_without_pca.predict(x_test)
```

```python
print("Model score with out pca is : ",r2_score(y_test,y_pred_without_pca)*100)
```

## with pca

```python
pca=PCA(n_components=0.80)
```

```python
x_train_pca=pca.fit_transform(x_train)
```

```python
x_test_pca=pca.transform(x_test)
```

```python
le=LinearRegression()
```

```python
start=time.time()
le.fit(x_train_pca,y_train,)
end=time.time()
print("The time required for traing is : ",start-end)
```

```python
y_pred_pca=le.predict(x_test_pca)
```

```python
print("Model score with out pca is : ",r2_score(y_test,y_pred_pca)*100)
```
