```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import DBSCAN
```

```python
df=pd.read_csv(r"C:\Users\CHAIT\OneDrive\Desktop\ML_git_hub\make_moons.csv")
```

```python
df.head()
```

```python
x=df.drop(columns=['True_Label'],axis=1)
```

```python
df.info()
```

```python
sns.scatterplot(x='Feature_1',y='Feature_2',data=df)
```

```python
db=DBSCAN(eps=0.2,min_samples=5,metric='euclidean')
```

```python
y_labels=db.fit_predict(x)
```

```python
sns.scatterplot(x='Feature_1',y='Feature_2',data=df,hue=y_labels)
```
