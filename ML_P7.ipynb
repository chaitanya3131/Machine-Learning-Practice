```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
```

```python
data=load_breast_cancer(as_frame=True)
```

```python
x=data.data
y=data.target
```

```python
sc=StandardScaler()
x=sc.fit_transform(x)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
linear_svm=SVC(C=1.0,kernel='linear',gamma='scale')
linear_svm.fit(x_train,y_train)

linear_svm_pred=linear_svm.predict(x_test)
```

```python
print('test acc = ',accuracy_score(y_test,linear_svm_pred)*100)

print('train acc = ',accuracy_score(y_train,linear_svm.predict(x_train))*100)
```

```python
nlinear_svm=SVC(C=1.0,kernel='rbf',gamma='scale')
nlinear_svm.fit(x_train,y_train)

nlinear_svm_pred=nlinear_svm.predict(x_test)
```

```python
print('test acc = ',accuracy_score(y_test,nlinear_svm_pred)*100)

print('train acc = ',accuracy_score(y_train,nlinear_svm.predict(x_train))*100)
```
