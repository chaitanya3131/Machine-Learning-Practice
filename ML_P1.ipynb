```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import StandardScaler,LabelEncoder,OneHotEncoder,OrdinalEncoder
```

```python
df=sns.load_dataset('titanic')
```

```python
df.info()
```

```python
df.head()
```

```python
df.describe()
```

```python
df=df.drop(columns=['deck','who'])
```

```python
print(df['embarked'].unique())
```

```python
df['age']=df['age'].fillna(df['age'].mean())
```

```python
df=df.dropna(how='any')
```

```python
df.head()
```

```python
lb=LabelEncoder()
df['sex']=lb.fit_transform(df['sex'])
```

```python
df['embarked']=lb.fit_transform(df['embarked'])
```

```python
one=OneHotEncoder(sparse_output=False)
```

```python
df_en=one.fit_transform(df[['embark_town']])
```

```python
df_en=pd.DataFrame(df_en,columns=one.get_feature_names_out(['embark_town']),index=df.index)
```

```python
df=df.drop(columns=['embark_town'],axis=1)
```

```python
df=pd.concat([df,df_en],axis=1)
```

```python
df['class'].unique()
```

```python
orr=OrdinalEncoder(handle_unknown='use_encoded_value',unknown_value=-1,categories=[['Third','Second','First']])
```

```python
df['class']=orr.fit_transform(df[['class']])
```

```python
df.head()
```

```python
df=df.dropna(how='any')
```

```python
df['alive']=df['alive'].map({'no':0,'yes':1})
```
