```python
df=sns.load_dataset('titanic')
df.head()
```

```python
df.drop(['sibsp','parch','embark_town','who','adult_male','deck','class'],axis=1,inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
df['sex']=le.fit_transform(df['sex'])
```

```python
df['embarked']=le.fit_transform(df['embarked'])
```

```python
df.head(2)
```

```python
df['alive']=df['alive'].map({'yes':1,'no':0})
```

```python
df['age']=df['age'].fillna(df['age'].mean())
df.info()
df.head()
```

```python
df=df.astype(int)
```

```python
df.head()
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
```

```python
nv=GaussianNB()
```

```python
x=df.drop(columns=['survived','alive'],axis=1)
y=df['survived']
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
nv.fit(x_train,y_train)
```

```python
y_pred_nv=nv.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred_nv)*100
```
