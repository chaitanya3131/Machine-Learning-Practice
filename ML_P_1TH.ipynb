```python
df=pd.read_csv(r"C:\Users\CHAIT\OneDrive\Desktop\ML_git_hub\titanic.csv")
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
```

```python
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df['Sex']=df['Sex'].map({'male':1,'female':0})
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
on=OneHotEncoder(sparse_output=False)
```

```python
df_en=on.fit_transform(df[['Embarked']])
```

```python
df_en=pd.DataFrame(df_en,columns=on.get_feature_names_out(['Embarked']),index=df.index)
```

```python
df_en
```

```python
df=df.drop(columns=['Embarked'],axis=1)
```

```python
df=pd.concat([df,df_en],axis=1)
```
